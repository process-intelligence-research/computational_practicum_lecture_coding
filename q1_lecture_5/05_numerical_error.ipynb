# Numerical error and step size

Compare the **forward Euler** solution against the exact solution for a simple first-order reaction, and see how the solution time scales with the time-step size.

## 1. Problem statement

We will use the reaction $A \overset{k}{\rightarrow} B$:

$$\frac{dC_A}{dt} = -kC_A, \qquad C_A(0)=1\ \mathrm{mol/L}, \qquad k=0.2\ \mathrm{min}^{-1}.$$

Assuming a constant total concentration of $1\ \mathrm{mol/L}$, $C_B = 1 - C_A$, $C_B(0)=0$.

The exact solutions are $C_A(t)=C_A(0)e^{-kt}$ and $C_B(t)=1-C_A(0)e^{-kt}$.

```python
# Import libraries
import time
from collections.abc import Callable
import matplotlib.pyplot as plt
import numpy as np

# Problem parameters
k = 0.2  # 1/min
c0 = np.array([1.0])  # mol/L
```

## 2. Reuse the forward Euler method

Reuse the `forward_euler` function defined in the previous example.

```python
def forward_euler(func: Callable, c0: np.ndarray, t: np.ndarray) -> np.ndarray:
    """Generic forward Euler method for initial value problem.

    Parameters
    ----------
    func : Callable
        ODE system to be solved
    c0 : np.ndarray
        Initial condition
    t : np.ndarray
        Time grid points

    Returns
    -------
    np.ndarray
        Solution of ODE system
    """
    # initialize arrays for time and solution values
    c = np.zeros((len(c0), len(t)))
    h = t[1] - t[0]

    # initial condition
    c[:, 0] = c0

    # iterate over each time step
    for i in range(1, len(t)):
        c[:, i] = c[:, i - 1] + h * func(t[i - 1], c[:, i - 1])
    return c
```

## 3. Define the ODE and the analytical solution

```python
# ODE definition
def dcA(t, c):
    return -k * c


# Analytical solution
def analytical_c(t):
    return c0[0] * np.exp(-k * t)
```

## 4. Compare a fine and a coarse time grid

Solve the IVP on both grids and measure how long each solve takes.

```python
# Define a fine and a coarse time grid
t_coarse = np.linspace(0, 30, 11)
t_fine = np.linspace(0, 30, 101)

# Forward Euler on the coarse grid
start_coarse = time.time()
c_coarse = forward_euler(dcA, c0, t_coarse)
time_coarse = time.time() - start_coarse
print(f"Euler method for coarse grid took {round(time_coarse, 4)} seconds.")

# Forward Euler on the fine grid
start_fine = time.time()
c_fine = forward_euler(dcA, c0, t_fine)
time_fine = time.time() - start_fine
print(f"Euler method for fine grid took {round(time_fine, 4)} seconds.")
```

## 5. Plot results

```python
fig, ax = plt.subplots()
ax.plot(t_coarse, c_coarse[0, :], "o-", label=f"C$_A$ (h=3.0 min)")
ax.plot(t_fine, c_fine[0, :], "s-", label=f"C$_A$ (h=0.3 min)")
ax.plot(t_fine, analytical_c(t_fine), "k--", label="C$_A$ - analytical solution")
ax.set(xlabel="Time [min]", ylabel="Concentration [mol/L]",
       title="Forward Euler: time-step size comparison")
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```
